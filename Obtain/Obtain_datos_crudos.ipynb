```python
import pandas as pd
import gspread

print("=" * 60)
print("INICIANDO INGESTA AUTOMATIZADA DE DATOS (OSEMN: OBTAIN)")
print("=" * 60)

try:
    # paso 1. Autenticación usando el archivo JSON en tu carpeta local
    print("🔑 Autenticando con Google Cloud...")
    cuenta_servicio = gspread.service_account(filename='credenciales_google.json')

    # paso 2. Conexión al archivo de Google Sheets (escribe el nombre exacto de tu archivo)
    print("🌐 Conectando al repositorio en Google Drive...")
    
    hoja_calculo = cuenta_servicio.open("Copia de respaldpo Instrumento De Recolección de datos STEAM de TIC COLEGIOS PEDRO VICENTE MALDONADO") 

    # paso 3. Seleccionar la primera pestaña donde caen los datos de Forms
    pestana = hoja_calculo.sheet1

    # paso 4. Extraer todos los registros y pasarlos directamente a Pandas
    print("📥 Descargando registros...")
    datos_crudos = pestana.get_all_records()
    df = pd.DataFrame(datos_crudos)

    # paso 5. Generar un respaldo físico local (Punto de control)
    ruta_respaldo = "encuestas_raw.csv"
    df.to_csv(ruta_respaldo, index=False, encoding='latin-1')

    print("=" * 60)
    print(f"✅ ¡Ingesta Exitosa! Se descargaron {len(df)} encuestas.")
    print(f"💾 Respaldo local guardado en: {ruta_respaldo}")
    print("=" * 60)
    
    display(df.head())

except Exception as e:
    print(f"❌ Error en la conexión: {e}")
    
# Verificación rápida del estado de los datos
    if not df.empty:
        print("✅ Confirmado: La conexión (Código 200) descargó los datos correctamente.")
        print(f"📊 Tienes un total de {len(df)} filas y {len(df.columns)} columnas.")
    display(df.head(5))
else:
    print("⚠️ El código 200 indica conexión exitosa, pero la hoja de cálculo parece estar vacía.")
```
