```python
        
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import textwrap
#from datetime import datetime

# --- 1. CONFIGURACIÓN ---
ARCHIVO_ENTRADA = "../Scrub/datos_Scrub.csv"

variables_numericas = ['edad'] 
variables_nominales = ['genero', 'institucion', 'area_residencia', 'idiomas']

variables_ordinales = [
    'año_escolar_num', 'edu_padres_num', 'ingreso_familiar_num', 'opinion_amigos_num',
    'nivel_informacion_steam_num', 'likert_apoyo_maestros', 'likert_apoyo_familia',
    'obstaculos_texto_num', 'sectores_interes_texto_num', 'likert_optimismo',
    'likert_estabilidad_eco', 'impacto_esperado_num', 'likert_percep_tecnologia',
    'area_tecnol_importante_num', 'likert_prob_steam', 'nivel_informacion_steam_num'
]

etiquetas_visuales = {
    "edad": "1: Edad del Estudiante",
    "genero": "2: Género del Estudiante", 
    "institucion": "3: Institución Educativa", 
    "area_residencia": "4: Área de Residencia", 
    "idiomas": "6: Idiomas que Hablan",
    "año_escolar_num": "5: Año Escolar Actual", 
    "edu_padres_num": "7: Nivel Educativo de los Padres",
    "ingreso_familiar_num": "8: Ingreso Familiar Mensual",
    "interes_cientifico_texto": "9: Actividades de interés",
    "actividades_gustan_texto": "10: Actividades que gusta realizar",
    "asignaturas_texto": "11: Asignaturas que más disfruta",
    "carreras_texto": "12: Carreras más atractivas",
    "influencia_carrera_texto": "13: Influencia en elección de carrera",
    "opinion_amigos_num": "15: Opinión de Amigos sobre STEAM",
    'obstaculos_texto_num': "17: Obstáculos en STEAM",
    'sectores_interes_texto_num': "18: Sectores de interés futuro",
    'impacto_esperado_num': "21: Impacto social esperado",
    'area_tecnol_importante_num': "23: Área tecnológica importante",
    "nivel_informacion_steam_num": "25: Nivel de Información STEAM",
    "likert_apoyo_maestros": "14: Apoyo de los Maestros",
    "likert_apoyo_familia": "16: Apoyo de la Familia",
    "likert_optimismo": "19: Optimismo sobre el Futuro",
    "likert_estabilidad_eco": "20: Estabilidad Económica",
    "likert_percep_tecnologia": "22: Percepción de la Tecnología",
    "likert_prob_steam": "Probabilidad de Elegir STEAM"
}

escala_ordinal_estricta = {
    "año_escolar_num": ['Primero', 'Segundo', 'Tercero de bachillerato'],
    "edu_padres_num": ['Primaria', 'Secundaria', 'Bachillerato', 'Universidad', 'Postgrado'],
    "ingreso_familiar_num": ['< $500', '$501 - $1000', '$1001 - $2000', '$2000 - $3000', '> $3000'],
    "opinion_amigos_num": ['Muy negativo', 'Negativo', 'Neutro', 'Positivo', 'Muy positivo'],
    "likert_apoyo_maestros": ['Nunca', 'Casi Nunca', 'A veces', 'Casi siempre', 'Siempre'],
    "likert_apoyo_familia": ['Nada', 'Poco', 'Moderadamente', 'Bastante', 'Mucho'],
    'obstaculos_texto_num': ['Recursos educativos', 'Apoyo familiar', 'Dificultad materias', 'Otro', 'No especifica'],
    'sectores_interes_texto_num': ['Salud', 'Tecnología', 'Medio ambiente', 'Arte', 'Educación (T/S/A)', 'No especifica'],
    "likert_optimismo": ['Nada optimista', 'Poco optimista', 'Neutral', 'Optimista', 'Muy optimista'],
    "likert_estabilidad_eco": ['Nada', 'Poco', 'Moderadamente', 'Bastante', 'Mucho'],
    "impacto_esperado_num":['Calidad de vida', 'Nuevas tecnologías','Sostenibilidad', 'Nuevas ideas artísticas','No estoy seguro', 'No especifica'],
    "likert_percep_tecnologia": ['Nada', 'Poco', 'Moderadamente', 'Bastante', 'Mucho'],
    "area_tecnol_importante_num": ['Salud', 'Energía', 'Educación', 'Transporte', 'Medio Ambiente'],
    "likert_prob_steam": ['Nada', 'Poco', 'Moderadamente', 'Bastante', 'Mucho'],
    'nivel_informacion_steam_num': ['Nada informado', 'Poco informado', 'Neutro', 'Informado', 'Muy informado']
}

preguntas_multiples_maestro = {
    "9: Actividades de Interés Científico y Tecnológico": {
        'q9_herramientas': 'Máquinas/equipos', 'q9_construccion_fisica': 'Construcción',
        'q9_mecanica': 'Mecánica/autos', 'q9_seres_vivos': 'Seres vivos',
        'q9_ingenieria': 'Ingeniería', 'q9_fenomenos_f': 'Fenómenos físicos',
        'q9_ajedrez': 'Ajedrez', 'q9_ciencia_ficcion': 'Ciencia ficción',
        'q9_misterios': 'Misterios', 'q9_logica_deductiva': 'Lógica deductiva',
        'q9_artes_plasticas': 'Artes plásticas', 'q9_arte_visual': 'Arte visual',
        'q9_actividades_artisticas': 'Artes escénicas'
    },
    "10: Actividades Prácticas": {
        'q10_descubrimientos_cient': 'Descubrimientos', 'q10_artes_visuales': 'Obras artísticas',
        'q10_explorar_tecnologia': 'Explorar tecnología', 'q10_analisis_datos': 'Analizar datos',
        'q10_problemas_software': 'Software', 'q10_problemas_creativos': 'Problemas creativos',
        'q10_problemas_matematicos': 'Matemáticas', 'q10_diseño_construccion': 'Diseño/Construcción',
        'q10_retroalim_arte': 'Retroalimentación arte', 'q10_estadistica': 'Estadística'
    },
    "11: Asignaturas": {
        'q11_matematicas': 'Matemáticas', 'q11_biologia': 'Biología', 'q11_fisica': 'Física',
        'q11_quimica': 'Química', 'q11_tecnologia': 'Tecnología', 'q11_arte': 'Arte'
    },
    "12: Carreras atractivas": {
        'q12_investigador': 'Investigación', 'q12_desarrollador': 'Desarrollo Software',
        'q12_ingeniero_civil': 'Ingeniería Civil', 'q12_artista_digital': 'Arte digital',
        'q12_matematico': 'Matemáticas/Estadística', 'q12_otro': 'Otras'
    }
}

# --- 2. INICIALIZACIÓN ---
sns.set_style("darkgrid", {"grid.color": "#929191", "grid.linestyle": "--"})
sns.set_palette("Set1")
sns.set_context("paper", font_scale=1.3)

df = pd.read_csv(ARCHIVO_ENTRADA, encoding='utf-8', sep=',', low_memory=False)
TOTAL = len(df)


def anotar_pct(ax, total):
    for p in ax.patches:
        if (h := p.get_height()) > 0:
            ax.annotate(f'{(h/total)*100:.1f}%', (p.get_x() + p.get_width() / 2., h),
                        ha='center', va='bottom', xytext=(0, 5), textcoords='offset points',
                        fontweight='bold', fontsize=12)
    plt.ylim(0, ax.get_ylim()[1] * 1.15)

# --- 3. VARIABLES NUMÉRICAS ---
for col in variables_numericas:
    if col in df.columns:
        tit = etiquetas_visuales.get(col, col)
        print(f"\n[NUMÉRICA] {tit}\nMedia: {df[col].mean():.2f} | Mediana: {df[col].median():.2f} | Moda: {df[col].mode()[0]} | SD: {df[col].std():.2f}")
        
        plt.figure(figsize=(8, 4))
        sns.histplot(data=df, x=col, kde=True, color='#2b83ba', bins=10)
        plt.axvline(df[col].mean(), color='red', linestyle='--', lw=2, label=f'Media: {df[col].mean():.1f}')
        plt.title(f'Distribución: {tit}', pad=10)
        plt.legend()
        plt.tight_layout()
        plt.show()

# --- 4. VARIABLES NOMINALES ---
for col in variables_nominales:
    if col in df.columns:
        tit = etiquetas_visuales.get(col, col)
        val_counts = df[col].value_counts()
        
        print(f"\n[NOMINAL] {tit}\nModa: {val_counts.index[0]}")
        res = pd.DataFrame({'N': val_counts, '%': (val_counts/TOTAL*100).round(1)})
        display(res)
        
        plt.figure(figsize=(9, 4))
        ax = sns.countplot(data=df, x=col, order=val_counts.index, palette="viridis")
        ax.set_xticklabels([textwrap.fill(t.get_text(), 15) for t in ax.get_xticklabels()])
        plt.title(tit, pad=10)
        anotar_pct(ax, TOTAL)
        plt.tight_layout()
        plt.show()

# --- 5. VARIABLES ORDINALES ---
for col in variables_ordinales:
    if col in df.columns:
        tit = etiquetas_visuales.get(col, col)
        etiquetas = escala_ordinal_estricta.get(col, None)
        df[col] = pd.to_numeric(df[col], errors='coerce')
        
        if etiquetas:
            frec_abs = df[col].value_counts().reindex(range(1, len(etiquetas) + 1), fill_value=0)
            frec_abs.index = etiquetas
        else:
            frec_abs = df[col].value_counts().sort_index()
            etiquetas = frec_abs.index.tolist()
            
        moda_idx = df[col].mode()[0]
        moda_txt = etiquetas[int(moda_idx)-1] if etiquetas and pd.notna(moda_idx) else moda_idx
        
        print(f"\n[ORDINAL] {tit}\nModa: {moda_txt}")
        res = pd.DataFrame({'N': frec_abs.values, '%': (frec_abs.values/TOTAL*100).round(1)}, index=frec_abs.index)
        display(res)
        
        plt.figure(figsize=(10, 4))
        ax = sns.barplot(x=frec_abs.index, y=frec_abs.values, palette="RdYlBu")
        ax.set_xticklabels([textwrap.fill(str(t), 12) for t in frec_abs.index], rotation=0)
        plt.title(tit, pad=10)
        anotar_pct(ax, TOTAL)
        plt.tight_layout()
        plt.show()

# --- 6. SELECCIÓN MÚLTIPLE ---
for tit, opciones in preguntas_multiples_maestro.items():
    cols_validas = [c for c in opciones.keys() if c in df.columns]
    
    if cols_validas:
        print(f"\n[MÚLTIPLE] {tit}")
        frec = df[cols_validas].sum().sort_values(ascending=False)
        
        res = pd.DataFrame({'N': frec, '%': (frec/TOTAL*100).round(1)})
        res.index = res.index.map(opciones)
        display(res)
        
        plt.figure(figsize=(10, 5))
        ax = sns.barplot(x=frec.values, y=res.index, palette="viridis")
        for i, v in enumerate(frec.values):
            ax.text(v + 0.5, i, str(v), va='center', fontweight='bold')
            
        plt.title(tit, pad=10)
        plt.xlim(0, frec.max() * 1.15)
        plt.tight_layout()
        plt.show()    
```
